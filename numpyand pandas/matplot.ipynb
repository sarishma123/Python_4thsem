```python
import numpy as np

x = np.linspace(-10, 10)

# 2. Define y values (y = x^2)
y = x**2

# 3. Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, y)

# Add labeling (optional)
plt.title("Basic Parabola:y=x**2")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()                # Display the graph
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create angles from 0 to 359 degrees
angle = np.arange(0, 360)

# 2. Convert angles to radians and calculate sin
y = np.sin(np.radians(angle))

# 3. Plot
plt.plot(angle, y)
plt.title("Sine curve")
plt.xlabel("Degrees")
plt.ylabel("sin(x)")
plt.show()
```

```python
plt.figure(figsize=(10,8))

# Subplot 1: Circle
plt.subplot(2,2,1)
t = np.linspace(0, 2*np.pi, 500)
plt.plot(np.cos(t), np.sin(t))
plt.title("Circle Graph")
plt.xlim(-1.5, 1.5)
plt.ylim(-1.1, 1.1)
plt.gca().set_aspect("equal")
plt.grid()

# Subplot 2: Parabola
plt.subplot(2,2,2)
y = np.linspace(-4, 4, 400)
x = y**2
plt.plot(x, y)
plt.plot(-x, y)
plt.title("Parabola Graph")
plt.xlim(-16, 16)
plt.ylim(-12, 12)
plt.grid()

# Subplot 3: Sin and Cos
plt.subplot(2,2,3)
a = np.arange(0,361)
r = np.radians(a)
plt.plot(a,np.sin(r))
plt.plot(a,np.cos(r))
plt.title("Sin and Cos graph")
plt.xlabel("Angle")
plt.ylabel("Value")
plt.xlim(0,360)
plt.ylim(-1.1,1.1)
plt.xticks([0,50,100,150,200,250,300,350])
plt.yticks([-1,-0.5,0,0.5,1])
plt.grid()

# Subplot 4: Tan
plt.subplot(2,2,4)
a = np.linspace(-360,360,3000)
r = np.radians(a)
t = np.tan(r)
# Don't filter large values
plt.plot(a,t)
plt.title("Tan Graph")
plt.xlabel("Angle")
plt.ylabel("Value")
plt.xlim(-360,360)
plt.ylim(-100,100)  # Increase range
plt.grid()

plt.tight_layout()
plt.show()
```
