```python
import matplotlib.pyplot as plt

# Data definition
flavors = ["Chocolate", "Vanilla", "Strawberry", "Mint", "Cookie Dough"]
votes = [8, 5, 3, 2, 6]
colors = ["chocolate", "gold", "pink", "lightgreen", "sandybrown"]

# Create the bar chart
bars = plt.bar(flavors, votes, color=colors)

# Add title and axis labels
plt.title("Favorite Ice Cream Flavors")
plt.xlabel("Flavors")
plt.ylabel("Votes")

# Add value labels on top of each bar
plt.bar_label(bars)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Task B: Histogram
ages = np.random.randint(18, 70, 50)
mean_age = np.mean(ages)

plt.figure()
plt.hist(ages, bins=10, edgecolor="black", color="skyblue")
plt.title("Age Distribution of Residents")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.axvline(
    mean_age,
    color="red",
    linestyle="--",
    linewidth=2,
)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Salary", "Rent", "Marketing", "IT"]

expenses = [45, 20, 15, 20]

colors = ["gold", "skyblue", "lightgreen", "tomato"]

plt.figure(figsize=(7, 7))

plt.pie(
    expenses,
    labels=labels,
    startangle=90,
    explode=[0.05, 0, 0, 0],
    colors=colors,
    shadow=True
)

plt.title("Company Expense Distribution")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(10)
data_group1 = np.random.normal(65, 8, 100)
data_group2 = np.random.normal(72, 10, 100)
data_group3 = np.random.normal(58, 6, 100)
data_group4 = np.random.normal(80, 12, 100)

plt.figure(figsize=(10, 6))
box = plt.boxplot([data_group1, data_group2, data_group3, data_group4],
                  patch_artist=True,
                  tick_labels=['Class A', 'Class B', 'Class C', 'Class D'],
                  boxprops=dict(facecolor='blue'),
                  whiskerprops=dict(color='black'),
                  capprops=dict(color='black'),
                  flierprops=dict(marker='o',
                                  markerfacecolor='red',
                                  markersize=6,
                                  alpha=0.5)
                 )

plt.title('Exam Score Distribution by Class', fontsize=16)
plt.ylabel('Exam Scores (out of 100)', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
