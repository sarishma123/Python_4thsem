```python
df["Unit_Price_NPR"] = pd.to_numeric(
    df["Unit_Price_NPR"], errors="coerce"
)

df["Unit_Price_NPR"] = df["Unit_Price_NPR"].fillna(
    df["Unit_Price_NPR"].median()
)

print(df["Unit_Price_NPR"])
```

```python
df["Discount"] = pd.to_numeric(
    df["Discount"], errors="coerce"
)

df["Discount"] = df["Discount"].fillna(
    df["Discount"].median()
)

print(df["Discount"])
```
