4.create matrix and uisng broadcasting,create a matrix where every elemnt is the sum of one value from A and one from B

```python
A=np.array([1,2,3])
B=np.array([10,20,30,40])
result=A[:,np.newaxis]+B
print("the result is:",result)
```
