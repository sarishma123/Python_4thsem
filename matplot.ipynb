```python
import numpy as np

x = np.linspace(-10, 10)

# 2. Define y values (y = x^2)
y = x**2

# 3. Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, y)

# Add labeling (optional)
plt.title("Basic Parabola:y=x**2")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()                # Display the graph
```

```python
plt.plot(x,y)
plt.title("square function")
plt.xlabel("x")
plt.ylabel("y=x**2")
plt.grid(True)
plt.show()
```

```python
plt.plot([0,0,100,100,0],[0,100,100,0,0])
plt.title("making box using matplotlib")
plt.axis([-10,110,-20,120])
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 7))

plt.plot([0, 100, 100, 0, 0], [0, 0, 100, 100, 0])
plt.plot([0, 50, 100], [100, 120, 100])
plt.plot([0, 100], [0, 100])
plt.plot([100, 0], [0, 100])

plt.title("Simple Envelope House")
plt.axis([-10, 110, -10, 130])
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create angles from 0 to 359 degrees
angle = np.arange(0, 360)

# 2. Convert angles to radians and calculate sin
y = np.sin(np.radians(angle))

# 3. Plot
plt.plot(angle, y)
plt.title("Sine curve")
plt.xlabel("Degrees")
plt.ylabel("sin(x)")
plt.show()
```
