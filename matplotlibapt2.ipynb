```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
angle = np.arange(0, 361)
y = np.sin(np.radians(angle))

fig, ax = plt.subplots()
ax.plot(angle, y)
ax.set_title("Sine Value")
ax.set_xlabel("Angle (degrees)")
ax.set_ylabel("Sine")
ax.grid(True)

plt.show()
```

```python
x = np.linspace(-1.4, 1.4, 30)

fig, ax = plt.subplots()
ax.plot(x, x**2)
fig.savefig(
    "my_square_function.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
x=np.linspace(-1.4, 1.4, 30)

fig, axs= plt.subplots(2,2,figsize=(5,5))
axs[0,0].plot(x,x)
axs[0,0].set_title("y=x")
axs[0,1].plot(x,x**2)
axs[0,1].set_title("y=x^2")
axs[1,0].plot(x,x**3)
axs[1,0].set_title("y=x^3")
axs[1,1].plot(x,x**4)
axs[1,1].set_title("y=x^4")

plt.tight_layout()
plt.show()
```
