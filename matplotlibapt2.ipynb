```python
import matplotlib.pyplot as plt
import numpy as np
```

Graph of sine

```python
angle = np.arange(0, 361)
y = np.sin(np.radians(angle))

fig, ax = plt.subplots()
ax.plot(angle, y)
ax.set_title("Sine Value")
ax.set_xlabel("Angle (degrees)")
ax.set_ylabel("Sine")
ax.grid(True)

plt.show()
```

Parbola Graph

```python
x = np.linspace(-1.4, 1.4, 30)

fig, ax = plt.subplots()
ax.plot(x, x**2)
fig.savefig(
    "my_square_function.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Line graph of y=x,x^2,x^3,x^4

```python
x=np.linspace(-1.4, 1.4, 30)

fig, axs= plt.subplots(2,2,figsize=(5,5))
axs[0,0].plot(x,x)
axs[0,0].set_title("y=x")
axs[0,1].plot(x,x**2)
axs[0,1].set_title("y=x^2")
axs[1,0].plot(x,x**3)
axs[1,0].set_title("y=x^3")

axs[1,1].plot(x,x**4)
axs[1,1].set_title("y=x^4")

plt.tight_layout()
plt.show()
```

```python


x = np.linspace(-1.4, 1.4, 30)

fig = plt.figure(figsize=(6, 4))
gs = fig.add_gridspec(2, 2)

axs1 = fig.add_subplot(gs[0, 0])
axs2 = fig.add_subplot(gs[0, 1])
axs3 = fig.add_subplot(gs[1, :])

axs1.plot(x, x)
axs1.set_title("y=x")

axs2.plot(x, x**2)
axs2.set_title("y=x^2")

axs3.plot(x, x**3)
axs3.set_title("y=x^3")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 11)  # Days 1 to 10
temperature = np.random.randint(15, 35, 10)  # Scale: 15-35 °C
revenue = np.random.randint(1000, 5000, 10)  # Scale: 1000-5000 $

fig, ax1 = plt.subplots(figsize=(10, 5))

# Left Y-axis (Revenue)
color1 = 'tab:blue'
ax1.set_xlabel('Day of Month')
ax1.set_ylabel('Revenue ($)', color=color1)
ax1.plot(x, revenue, marker='s', color=color1, linestyle='--', linewidth=2, label='Revenue')
ax1.tick_params(axis='y', labelcolor=color1)
ax1.grid(alpha=0.2)

# Right Y-axis (Temperature)
ax2 = ax1.twinx()
color2 = 'tab:red'
ax2.set_ylabel('Temperature (°C)', color=color2)
ax2.plot(x, temperature, marker='o', color=color2, linewidth=2, label='Temp')
ax2.tick_params(axis='y', labelcolor=color2)

# Title and Legend
plt.title("Daily Revenue vs. Temperature (Dual Axis)", fontsize=16)
fig.legend(loc='upper left', bbox_to_anchor=(0.1, 0.9))
plt.tight_layout()

plt.show()
```

```python
x=np.linspace(-1.4, 1.4, 30)
plt.plot(x,x**2,"r--",label="square function")
plt.plot(x,x**3,"g--",label="cube function")
plt.legend(loc="best")
plt.grid()
plt.show()
```

```python
x = np.linspace(0.1, 15, 500)
y = x**3/np.exp(2*x)

plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.plot(x, y)
plt.yscale('linear')
plt.title('linear')
plt.grid(True)

plt.subplot(2,2,2)
plt.plot(x, y)
plt.yscale('log')
plt.title('log')
plt.grid(True)

plt.subplot(2,2,3)
plt.plot(x, y)
plt.yscale('logit')
plt.title('logit')
plt.grid(True)

plt.subplot(2,2,4)
plt.plot(x, y - y.mean())
plt.yscale('symlog', linthresh=0.05)
plt.title('symlog')
plt.grid(True)

plt.show()
```
